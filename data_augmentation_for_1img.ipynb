```python
from keras.preprocessing import image
from keras.preprocessing.image import ImageDataGenerator
```

```python
img=image.load_img('path',target_size=('your target size'))
```

```python
plt.imshow(img)
```

```python
# here we are deciding ki kya transformation ho raha hai

datagen=ImageDataGenerator(rotation_range=30 ,
                           shear_range=0.2 ,
                           zoom_range=0.2  ,
                           horizontal_flip=True,
                           vertical_flip=True,
                           width_shift_range=0.2,
                           height_shift_range=0.2,
                           fill_mode='reflect' # or constant -> there are multiple fill use any of them .
)
```

```python
img=image.img_to_array(img)
```

```python
img.shape
```

```python
type(img)
```

```python
input_batch=img.reshape(1,200,200,3)
```

```python
i=0

# flow is used for single image
for output in datagen.flow(input_batch,batch_size=1,save_to_dir='aug'):
  i=i+1

  if i==10:
    break
```
